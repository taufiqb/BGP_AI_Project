```python
import pybgpstream
import time
from datetime import datetime

# --------------------------------------------------
# Configuration
# --------------------------------------------------

TEST_DURATION = 600  # 10 minutes

# --------------------------------------------------
# Connect to Route Views stream
# --------------------------------------------------

stream = pybgpstream.BGPStream(
    project="routeviews-stream"
)

print("Connected to Route Views BMP stream")
print("Starting 10-minute BGP streaming test...")
print()

# --------------------------------------------------
# Streaming test
# --------------------------------------------------

start_time = time.monotonic()
start_datetime = datetime.now()

count = 0
last_report = start_time

for elem in stream:

    count += 1

    # Current elapsed time
    elapsed = time.monotonic() - start_time

    # Display first few elements
    if count <= 5:
        print(elem)

    # Progress report every 30 seconds
    if time.monotonic() - last_report >= 30:

        print(
            f"[{datetime.now().strftime('%H:%M:%S')}] "
            f"Elapsed: {elapsed:.0f}s | "
            f"BGP elements received: {count}"
        )

        last_report = time.monotonic()

    # Stop after 10 minutes
    if elapsed >= TEST_DURATION:
        break


# --------------------------------------------------
# Test summary
# --------------------------------------------------

end_datetime = datetime.now()
total_time = time.monotonic() - start_time

print()
print("=" * 60)
print("10-MINUTE BGP STREAMING TEST COMPLETED")
print("=" * 60)

print(f"Start time       : {start_datetime}")
print(f"End time         : {end_datetime}")
print(f"Duration         : {total_time:.1f} seconds")
print(f"BGP elements     : {count}")
print(f"Average rate     : {count / total_time:.2f} elements/sec")
print("=" * 60)
```
